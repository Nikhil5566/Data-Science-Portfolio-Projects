```python
import math

# --- 1. Numerical Columns (Histograms in a Grid) ---
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

if len(numerical_cols) > 0:
    # Defining grid size
    num_plots = len(numerical_cols)
    num_cols = 3
    num_rows = math.ceil(num_plots / num_cols)

    # Creating figure and axes
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(5 * num_cols, 4 * num_rows))
    axes = axes.flatten()

    # Iterating and plot
    for i, col in enumerate(numerical_cols):
        sns.histplot(data=df, x=col, kde=True, ax=axes[i], color='teal') # Added color
        axes[i].set_title(f'Distribution of {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Frequency')

    # Hiding any unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns found.")


# --- 2. Categorical Columns (Countplots in a Grid) ---
# Filtering out columns with too many unique values (like IDs) to avoid messy plots
categorical_cols = [c for c in df.select_dtypes(include='object').columns if df[c].nunique() < 20]

if len(categorical_cols) > 0:
    # Defining grid size
    num_plots = len(categorical_cols)
    num_cols = 3
    num_rows = math.ceil(num_plots / num_cols)

    # Creating figure and axes
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(5 * num_cols, 4 * num_rows))
    axes = axes.flatten()

    # Iterating and plot
    for i, col in enumerate(categorical_cols):
        sns.countplot(data=df, x=col, ax=axes[i], palette='viridis', hue=col, legend=False) # Added palette
        axes[i].set_title(f'Distribution of {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Count')
        axes[i].tick_params(axis='x', rotation=45)

    # Hiding unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()

     # --- 3. Value Counts (Text Output) ---
    print("--- Categorical Value Counts ---")
    for col in categorical_cols:
        print(f"\nValue counts for {col}:")
        print(df[col].value_counts())
else:
    print("No categorical columns found (or all have high cardinality).")
```

# Bivariate Analysis

```python
import itertools

# --- 1. Numerical-Numerical Relationships (Scatter Plots) ---
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Generating all unique pairs of numerical columns
num_pairs = list(itertools.combinations(numerical_cols, 2))

if len(num_pairs) > 0:
    # Defining grid size
    num_plots = len(num_pairs)
    num_cols = 3
    num_rows = math.ceil(num_plots / num_cols)

    # Creating figure and axes
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(5 * num_cols, 4 * num_rows))

    if num_plots == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    # Iterating through pairs and plot
    for i, (col1, col2) in enumerate(num_pairs):
        sns.scatterplot(data=df, x=col1, y=col2, ax=axes[i], alpha=0.5, color='crimson') # Added color
        axes[i].set_title(f'{col1} vs {col2}')
        axes[i].set_xlabel(col1)
        axes[i].set_ylabel(col2)

    # Hiding unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Not enough numerical columns for scatter plots.")


# --- 2. Categorical-Numerical Relationships (Box Plots) ---
# Excluding high cardinality columns
categorical_cols = [c for c in df.select_dtypes(include='object').columns if df[c].nunique() < 20]
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Generating all pairs of (Categorical, Numerical)
cat_num_pairs = list(itertools.product(categorical_cols, numerical_cols))

if len(cat_num_pairs) > 0:
    # Defining grid size
    num_plots = len(cat_num_pairs)
    num_cols = 3
    num_rows = math.ceil(num_plots / num_cols)

    # Creating figure and axes
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(5 * num_cols, 4 * num_rows))

    if num_plots == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    # Iterating through pairs and plot
    for i, (cat_col, num_col) in enumerate(cat_num_pairs):
        sns.boxplot(data=df, x=cat_col, y=num_col, ax=axes[i], palette='Set2', hue=cat_col, legend=False)
        axes[i].set_title(f'{num_col} by {cat_col}')
        axes[i].set_xlabel(cat_col)
        axes[i].set_ylabel(num_col)
        axes[i].tick_params(axis='x', rotation=45)

    # Hiding unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("No categorical-numerical pairs found.")


# --- 3. Categorical-Categorical Relationships (Heatmaps) ---
# Generating pairs of categorical columns
cat_pairs = list(itertools.combinations(categorical_cols, 2))

if len(cat_pairs) > 0:
    num_plots = len(cat_pairs)
    num_cols = 3
    num_rows = math.ceil(num_plots / num_cols)

    fig, axes = plt.subplots(num_rows, num_cols, figsize=(5 * num_cols, 4 * num_rows))

    if num_plots == 1:
        axes = [axes]
    else:
        axes = axes.flatten()

    for i, (col1, col2) in enumerate(cat_pairs):
        # Creating a cross-tabulation
        crosstab = pd.crosstab(df[col1], df[col2], normalize='index')
        sns.heatmap(crosstab, annot=True, fmt='.2f', cmap='Purples', ax=axes[i])
        axes[i].set_title(f'{col1} vs {col2} (Proportion)')

    for j in range(i + 1, len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
```

# Statistical Analysis

## Gaming Metrics and Feature Engineeing

```python
import scipy.stats as stats

print("--- 1. ENGINEERING SENTIMENT & RETENTION METRICS ---")
# Here we analyze games properly, raw positive/negative review counts are useless.
# We must engineer 'Approval Rating' and estimate the 'Total Player Base'.

# Assuming columns: 'positive', 'negative', 'price', 'average_forever', 'median_forever', 'owners'
df_clean = df.dropna(subset=['positive', 'negative', 'price', 'median_forever']).copy()

# 1. Total Engagement & Approval Rating
df_clean['Total_Reviews'] = df_clean['positive'] + df_clean['negative']
# Filterring out games with less than 100 reviews to remove obscure, unplayed noise
df_clean = df_clean[df_clean['Total_Reviews'] > 100]
df_clean['Approval_Rating_%'] = (df_clean['positive'] / df_clean['Total_Reviews']) * 100

# 2. Parsing the 'Owners' Column (SteamSpy usually formats this as '1,000,000 .. 2,000,000')
if df_clean['owners'].dtype == 'O':
    # Extracting the lower bound of the owners tier for mathematical sorting
    df_clean['Estimated_Owners_Base'] = df_clean['owners'].str.split(' .. ').str[0].str.replace(',', '').astype(float)
else:
    df_clean['Estimated_Owners_Base'] = df_clean['owners']

# 3. Categorizing Pricing Tiers for Market Analysis
conditions = [
    (df_clean['price'] == 0),
    (df_clean['price'] > 0) & (df_clean['price'] <= 1500), # Assuming price is in cents ($15) or adjust if it's in dollars
    (df_clean['price'] > 1500) & (df_clean['price'] <= 4000),
    (df_clean['price'] > 4000)
]
choices = ['Free-to-Play', 'Indie (<$15)', 'Mid-Tier ($15-$40)', 'AAA (>$40)']
df_clean['Pricing_Tier'] = np.select(conditions, choices, default='Unknown')

print(" Engineered: 'Approval_Rating_%', 'Estimated_Owners_Base', and 'Pricing_Tier'.")

# :::::: BEHAVIORAL HYPOTHESIS TESTING :::::::

print("\n--- 2. THE SUNK-COST FALLACY (KRUSKAL-WALLIS H-TEST) ---")
# Hypothesis: Does paying a premium price for a game result in a higher median playtime?
# We use Kruskal-Wallis because playtime is heavily right-skewed (violating ANOVA normality).

tiers = ['Free-to-Play', 'Indie (<$15)', 'Mid-Tier ($15-$40)', 'AAA (>$40)']
# Ensure data exists for these tiers
tier_groups = [df_clean[df_clean['Pricing_Tier'] == tier]['median_forever'].dropna() for tier in tiers if len(df_clean[df_clean['Pricing_Tier'] == tier]) > 50]

if len(tier_groups) > 1:
    h_stat, p_val_kw = stats.kruskal(*tier_groups)

    print(f"Kruskal-Wallis P-Value (Playtime Variance across Price): {p_val_kw:.4e}")
    if p_val_kw < 0.05:
        print(" Insight: Statistically significant variance in playtime across pricing tiers.")
        print(" -> The 'Sunk-Cost Fallacy' or 'F2P Grind' effect is real. Pricing fundamentally dictates how long a player stays engaged.")
    else:
        print(" Insight: Playtime is independent of price. A $60 game holds attention exactly as well as a Free game.")

    # Visualizing the Playtime Economics
    plt.figure(figsize=(10, 6))
    # Using log-scale for playtime because a few games (like CS:GO/Dota) have thousands of hours, ruining the chart scale
    sns.boxplot(x='Pricing_Tier', y=np.log1p(df_clean['median_forever']), data=df_clean, order=tiers, palette='magma')
    plt.title('Player Retention Economics: Log(Median Playtime) by Pricing Tier', fontsize=14, fontweight='bold')
    plt.xlabel('Market Pricing Tier')
    plt.ylabel('Natural Log of Median Playtime (Minutes)')
    plt.show()


print("\n--- 3. THE MAINSTREAM TOXICITY PARADOX (SPEARMAN RANK CORRELATION) ---")
# Do games with massive player bases (high Total Reviews) suffer from lower Approval Ratings?

spearman_coef, p_val_spearman = stats.spearmanr(df_clean['Total_Reviews'], df_clean['Approval_Rating_%'])

print(f"Spearman Correlation Coefficient (Popularity vs. Approval): {spearman_coef:.4f}")
print(f"P-Value: {p_val_spearman:.4e}")

if p_val_spearman < 0.05:
    if spearman_coef < 0:
        print(" Insight: Negative correlation detected. The 'Toxicity Paradox' is mathematically real.")
        print("   -> As a game becomes mainstream, its approval rating statistically degrades. Niche games have happier, protective communities.")
    else:
        print(" Insight: Positive correlation detected. Masterpieces gain both massive player bases AND high ratings.")
else:
    print(" Insight: Popularity has zero mathematical impact on community sentiment.")

# Visualizing the Toxicity Paradox
plt.figure(figsize=(10, 6))
sns.regplot(x=np.log1p(df_clean['Total_Reviews']), y=df_clean['Approval_Rating_%'],
            scatter_kws={'alpha':0.2, 'color':'purple'}, line_kws={'color':'gold', 'linewidth':2})
plt.title('Community Psychology: Log(Total Reviews) vs. Approval Rating', fontsize=14, fontweight='bold')
plt.xlabel('Natural Log of Total Reviews (Scale of Player Base)')
plt.ylabel('Approval Rating (%)')
plt.show()


print("\n--- 4. THE PARETO PRINCIPLE (POWER LAW OF PLAYER ATTENTION) ---")
# 80% of all playtime and reviews on Steam are monopolized by the top 20% of games.
# This proves whether Steam is a healthy middle-class marketplace or a heavily monopolized oligopoly.

# Sorting games by Total Reviews (as a proxy for engagement/sales)
pareto_df = df_clean.sort_values(by='Total_Reviews', ascending=False).reset_index(drop=True)

# Calculating cumulative percentages
pareto_df['Cumulative_Reviews'] = pareto_df['Total_Reviews'].cumsum()
total_steam_reviews = pareto_df['Total_Reviews'].sum()
pareto_df['Cumulative_Percentage'] = (pareto_df['Cumulative_Reviews'] / total_steam_reviews) * 100

# Finding what percentage of games hold 80% of the engagement
top_games_count = len(pareto_df[pareto_df['Cumulative_Percentage'] <= 80])
total_games = len(pareto_df)
percentage_of_games = (top_games_count / total_games) * 100

print(f" PARETO DISTRIBUTION DETECTED:")
print(f" -> Just {percentage_of_games:.2f}% of the games on Steam ({top_games_count} titles) control exactly 80.00% of all player engagement.")
if percentage_of_games < 20:
    print("   -> Conclusion: Steam is a hyper-monopolized 'Winner-Takes-All' market. It is severely harsher than the standard 80/20 rule.")

# Visualizing the Monopoly
plt.figure(figsize=(10, 6))
plt.plot((pareto_df.index / total_games) * 100, pareto_df['Cumulative_Percentage'], color='red', linewidth=3)
plt.axvline(percentage_of_games, color='black', linestyle='--', label=f'Top {percentage_of_games:.1f}% of Games')
plt.axhline(80, color='blue', linestyle='--', label='80% of Total Engagement')
plt.fill_between((pareto_df.index / total_games) * 100, pareto_df['Cumulative_Percentage'], color='red', alpha=0.1)
plt.title('The Winner-Takes-All Market: Cumulative Engagement Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Percentage of Total Games on Steam (%)')
plt.ylabel('Cumulative Player Engagement (%)')
plt.legend(loc='lower right')
plt.grid(True, alpha=0.3)
plt.show()
```

# Feature Engineering

```python
from sklearn.preprocessing import StandardScaler

print("--- 1. TARGET ISOLATION & DATA CLEANING ---")
# We are building a Regressor to predict the game's 'Approval_Rating_%'.
# We drop any rows where our core predictive features are missing.

ml_df = df_clean.dropna(subset=['Approval_Rating_%', 'price', 'Estimated_Owners_Base', 'median_forever', 'Total_Reviews']).copy()
y = ml_df['Approval_Rating_%']

print(" Target Formulated: 'Approval_Rating_%' isolated.")


print("\n--- 2. LOW-CARDINALITY ONE-HOT VECTORIZATION ---")
# Algorithms cannot read "Free-to-Play". We must convert our Pricing Tiers into binary vectors (0 or 1).
# We use 'drop_first=True' to avoid perfect multicollinearity (The Dummy Variable Trap).

ml_df_encoded = pd.get_dummies(ml_df, columns=['Pricing_Tier'], drop_first=True)

print(" Engineered: Monetization tiers projected into sparse binary vectors.")


print("\n--- 3. CONTINUOUS VARIANCE STANDARDIZATION ---")
# Gradient-based algorithms are highly sensitive to spatial magnitude.
# 'Estimated_Owners_Base' (e.g., 5,000,000) will mathematically overpower 'price' (e.g., 60) if unscaled.

continuous_features = ['price', 'Estimated_Owners_Base', 'median_forever', 'Total_Reviews']

# Initialize the Standard Scaler
scaler = StandardScaler()

# We only scale the continuous features
ml_df_encoded[continuous_features] = scaler.fit_transform(ml_df_encoded[continuous_features])

# Define the final Matrix (X), dropping the target and any remaining text columns
# We dynamically select our scaled features and our new dummy variables
feature_columns = continuous_features + [col for col in ml_df_encoded.columns if 'Pricing_Tier_' in col]
X_adv = ml_df_encoded[feature_columns].copy()

print(f" Final Machine-Readable Matrix Assembled: {X_adv.shape[1]} predictive dimensions.")
print("\n Final Matrix (X_adv) Preview:")
print(X_adv.head(3))
```

# Algo Hit Prediction

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


print("--- 1. OUT-OF-SAMPLE DATA SPLITTING ---")
# We split the data into 80% Training and 20% Out-of-Sample Testing.
# The algorithm will learn the psychology of the market on the 80%, and we will test its actual predictive power on the unseen 20%.

X_train, X_test, y_train, y_test = train_test_split(X_adv, y, test_size=0.2, random_state=42)

print(f" Training Set: {len(X_train):,} games")
print(f" Testing Set: {len(X_test):,} games (Strictly Unseen)")


print("\n--- 2. ALGORITHMIC TRAINING (BAGGING VS. BOOSTING) ---")
# 1. Random Forest Regressor
# We restrict max_depth to prevent the model from memorizing the training data.
rf_model = RandomForestRegressor(n_estimators=200, max_depth=8, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

# 2. XGBoost Regressor
# We use a lower learning rate (eta) to ensure the gradient descent converges smoothly.
xgb_model = XGBRegressor(n_estimators=200, max_depth=6, learning_rate=0.05, random_state=42, n_jobs=-1)
xgb_model.fit(X_train, y_train)

print(" Both Tree-Based Regression Architectures successfully trained.")


print("\n--- 3. OUT-OF-SAMPLE EVALUATION METRICS ---")
# Force both models to predict the approval ratings of the unseen 20% of the market.
rf_preds = rf_model.predict(X_test)
xgb_preds = xgb_model.predict(X_test)

def evaluate_model(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f" {name} Performance:")
    print(f"   -> Mean Absolute Error (MAE): {mae:.2f}% Rating Points")
    print(f"   -> Root Mean Squared Error (RMSE): {rmse:.2f}% Rating Points")
    print(f"   -> R-Squared (R2): {r2:.4f}")
    return r2, mae

r2_rf, mae_rf = evaluate_model("Random Forest", y_test, rf_preds)
print("-" * 40)
r2_xgb, mae_xgb = evaluate_model("XGBoost", y_test, xgb_preds)

# Determining the superior architecture
best_model_name = "XGBoost" if r2_xgb > r2_rf else "Random Forest"
best_preds = xgb_preds if r2_xgb > r2_rf else rf_preds
best_model = xgb_model if r2_xgb > r2_rf else rf_model

print(f"\n The {best_model_name} mathematically outperformed the alternative in predictive accuracy.")


print("\n--- 4. VISUALIZING PREDICTIVE ACCURACY ---")
# Here we are plotting the Actual Approval Ratings vs. the Algorithm's Predicted Ratings.

plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test, y=best_preds, alpha=0.4, color='cyan', edgecolor='black', s=40)

# Plotting the Line of Perfect Prediction
min_val = min(y_test.min(), best_preds.min())
max_val = max(y_test.max(), best_preds.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Perfect Prediction Diagonal')

plt.title(f'Algorithmic Accuracy: Actual vs. Predicted Community Rating ({best_model_name})', fontsize=14, fontweight='bold')
plt.xlabel('Actual Approval Rating (%)', fontsize=12)
plt.ylabel('Algorithm Predicted Rating (%)', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("\n--- 5. ALGORITHMIC FEATURE IMPORTANCE ---")

importances = pd.Series(best_model.feature_importances_, index=X_adv.columns).sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=importances.values, y=importances.index, palette='mako')
plt.title(f'Algorithmic Drivers: Top Features Dictating a Game\'s Success ({best_model_name})', fontsize=14, fontweight='bold')
plt.xlabel('Algorithmic Importance Weight (Gini Reduction)', fontsize=12)
plt.ylabel('Engineered Feature', fontsize=12)
plt.show()
```
